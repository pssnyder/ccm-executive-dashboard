## Visualizations

```python
# Plot cash over time
balance_sheet['Cash_num'] = pd.to_numeric(balance_sheet['Cash'], errors='coerce')

plt.figure(figsize=(14, 6))
plt.plot(balance_sheet['Timestamp'], balance_sheet['Cash_num'], marker='o', linewidth=2)
plt.title('Cash Balance Over Time', fontsize=16)
plt.xlabel('Date')
plt.ylabel('Cash ($)')
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Economic phase distribution
phase_counts = economic_phases['game.phase'].value_counts()

plt.figure(figsize=(10, 6))
colors = {'recession': '#ef4444', 'normal': '#eab308', 'boom': '#22c55e'}
phase_counts.plot(kind='bar', color=[colors.get(p, '#888') for p in phase_counts.index])
plt.title('Economic Phase Distribution', fontsize=16)
plt.xlabel('Phase')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print(phase_counts)
```
