```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sqlalchemy import create_engine

# Set style
sns.set_style('darkgrid')
plt.rcParams['figure.figsize'] = (12, 6)

# Database connection
engine = create_engine('postgresql://postgres:postgres@postgres:5432/nocodb')
```

## Load Data

```python
# Load key datasets
balance_sheet = pd.read_sql('SELECT * FROM balance_sheet', engine)
income_stmt = pd.read_sql('SELECT * FROM income_statement', engine)
economic_phases = pd.read_sql('SELECT * FROM economic_phases', engine)

print(f"Balance Sheet: {len(balance_sheet)} records")
print(f"Income Statement: {len(income_stmt)} records")
print(f"Economic Phases: {len(economic_phases)} records")
```

## Balance Sheet Analysis

```python
# Display recent balance sheet
balance_sheet.head(10)
```

```python
# Plot cash trend
plt.figure(figsize=(14, 6))
plt.plot(balance_sheet['timestamp'], balance_sheet['cash'].astype(float), marker='o')
plt.title('Cash Balance Over Time')
plt.xlabel('Date')
plt.ylabel('Cash ($)')
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Economic Cycle Analysis

```python
# Analyze phase distribution
phase_counts = economic_phases['game_phase'].value_counts()

plt.figure(figsize=(10, 6))
phase_counts.plot(kind='bar', color=['#ef4444', '#eab308', '#22c55e'])
plt.title('Economic Phase Distribution')
plt.xlabel('Phase')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

print(f"\nPhase Statistics:")
print(phase_counts)
```
